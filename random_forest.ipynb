```python
# 1. Sample delivery data
import numpy as np
rng = np.random.default_rng(42)
km = rng.integers(2, 41, 400)
stops = rng.integers(1, 16, 400)
rain = rng.integers(0, 2, 400)
X = np.column_stack([km, stops, rain])
#print(X[:4])
import pandas as pd
df = pd.DataFrame(X, columns=["km", "stops", "rain"])
df.head(7)
```

```python
minutes = 12 + 1.2 * km + 2.5 * stops
minutes = minutes + 14 * rain
minutes = minutes + rng.normal(0, 10, len(X))
y = (minutes > 65).astype(int)
print(np.bincount(y))
```

```python
print(minutes)
print(y)
print(np.bincount(y))
```

```python
# 3. Split the data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)
print(len(y_train), len(y_test))
```

```python
# 4. Train the Random Forest
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier(
    n_estimators=100, max_depth=5, random_state=42
)
forest.fit(X_train, y_train)
pred = forest.predict(X_test)
```

```python
# 5. Check the test results
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

print(f"Accuracy: {accuracy_score(y_test, pred):.2%}")
print(confusion_matrix(y_test, pred, labels=[0, 1]))
```

```python
# 6. Try the same route in dry weather and rain
routes = [[22, 8, 0], [22, 8, 1]]  # new data(routes)

# Bashorat qilish (0 = on time, 1 = late)
predictions = forest.predict(routes)
probabilities = forest.predict_proba(routes)

labels = ["on time", "late"]

for route, pred, prob in zip(routes, predictions, probabilities):
    weather = "Rain" if route[2] == 1 else "Dry weather"
    print(f"Route {route} ({weather}) -> Result: {labels[pred]} (Probability of late: {prob[1]:.2%})")
```
