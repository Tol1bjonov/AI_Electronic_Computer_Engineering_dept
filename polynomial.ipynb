```python
#  polynomial
import matplotlib.pyplot as plt
x = [1,2,3,5,6,7,8,9,10,12,13,14,15,16,18,19,21,22]
y = [100,90,80,60,60,55,60,65,70,70,75,76,78,79,90,99,99,100]
plt.scatter(x, y)
plt.show()
```

```python
#######
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract the coefficients in the form f(x) = ax^2 + bx + c
a, b, c = mymodel.coefficients

# Define the prediction function
def f(x_value):
    return mymodel(x_value)

# Print the prediction equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Generate x-values for prediction
myline = np.linspace(1, 22, 100)

# Visualization
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```

```python
# Predict speed at 17:00
time = 17
predicted_speed = f(time)

# 1. Print predicted speed
print(f"Predicted speed at 17:00 = {predicted_speed:.2f}")

# 2. Display predicted value on the graph
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')

# Predicted point
plt.scatter(time, predicted_speed, color='green', s=100)
plt.annotate(
    f"{predicted_speed:.2f}",
    (time, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points"
)

plt.title("YUSUFJON: Polynomial Regression (degree=2)")
plt.xlabel("Time")
plt.ylabel("Speed")
plt.show()
```
